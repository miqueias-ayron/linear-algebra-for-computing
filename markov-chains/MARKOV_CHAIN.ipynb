# Markov Chains


```python
import numpy as np

P = np.array([[0.7, 0.2, 0.1],
              [0.3, 0.4, 0.3],
              [0.2, 0.4, 0.4]])

pi_0 = np.array([1,0,0])
print(pi_0 @ np.linalg.matrix_power(P,2))
print(pi_0 @ np.linalg.matrix_power(P,100_000))
```

```python
rng = np.random.default_rng()
estado, estados = 0, ["Sol", "Nublado", "Chuva"]

for _ in range(10):
    estado = rng.choice(3, p=P[estado]) 
    print(estados[estado], end=" ")
```

```python
contagem = np.zeros(3)
print(contagem)
estado = 0

for _ in range(100_000):
    estado = rng.choice(3, p=P[estado]) 
    contagem[estado] += 1

print(contagem/contagem.sum())
```
